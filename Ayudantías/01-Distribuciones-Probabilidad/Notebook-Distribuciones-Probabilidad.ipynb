# Distribuciones de Probabilidad en Variables Aleatorias Continuas

Este es un notebook orientado a facilitar la comprensión de las distribuciones de probabilidad.

Acá encontrarán gráficos interactivos para cada distribución, además de una breve explicación de cada coeficiente de la distribución.

# 1. Importar librerias requeridas

```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
```

# 2. Distribución uniforme

Es una de las distribuciones más sencillas, donde la probabilidad es constante en todo el dominio del experimento. En estos contextos,

```python
def dist_uniforme(a:float, b:float):
    x = np.linspace(0, 10, 1000)

    if a > b: #Swap debido a que se requiere que a < b
        print("Se requiere a < b, se intercambiaron los valores")
        temp = a
        a = b
        b = temp
    elif a == b:
        print("Error: a y b no pueden ser iguales.")
        return
    y = np.where((x >= a) & (x <= b), 1 / (b - a), 0)
    plt.plot(x, y)
    plt.title(f"Distribución uniforme con a={a} y b={b}")
    plt.grid(True)
    plt.show()

widgets.interact(
    dist_uniforme,
    a=widgets.FloatSlider(value=0,min=0,max=10, step= 0.5),
    b=widgets.FloatSlider(value=1,min=0,max=10, step= 0.5),
)
```

# 3. Distribución exponencial

Esta distribución es muy utilizada para modelar decrecimientos, desgastes o experimentos donde es relevante el tiempo transcurrido hasta la ocurrencia del evento.

Esta distribución utiliza como parámetro el coeficiente $\lambda$, que corresponde a la tasa de ocurrencia de eventos por unidad de tiempo.

Ejemplo: Una tasa de 3 fallos al año en un sistema crítico implicaría un $\lambda = 3 \left[\frac{falla}{año}\right]$

```python
def dist_exp(lambda_coef:float):
    x = np.linspace(0, 10, 1000)
    y = lambda_coef * np.exp(-lambda_coef*x)

    plt.plot(x, y)
    plt.title(f"Distribución exponencial con lambda={lambda_coef}")
    plt.grid()
    plt.show()

widgets.interact(dist_exp,
                 lambda_coef=widgets.FloatSlider(value=-0.1,min=0.1,max=10, step= 0.25))
```

# 4. Distribución Normal

Esta distribución es muy utilizada para modelar fenómenos donde los valores tienden a concentrarse alrededor de un valor medio, presentando variaciones simétricas respecto a este.

Esta distribución utiliza como parámetros la media $\mu$, que corresponde al valor esperado de la variable aleatoria, y la desviación estándar $\sigma$, que determina qué tan dispersos se encuentran los valores respecto a la media.

Ejemplo: Si la duración promedio de una batería es de 1000 horas con una desviación estándar de 50 horas, entonces $\mu = 1000\ [h]$ y $\sigma = 50\ [h]$.

```python
def dist_normal(mu: float, sigma: float):
    if sigma <= 0:
        print("Error: sigma debe ser mayor que 0.")
        return

    # Rango centrado en la media
    x = np.linspace(-10, 10, 10000)

    # Función de densidad de probabilidad
    y = (
        1 / (sigma * np.sqrt(2 * np.pi))
        * np.exp(-((x - mu)**2) / (2 * sigma**2))
    )

    plt.plot(x, y)


    plt.xlabel("x")
    plt.ylabel("f(x)")
    plt.title(
        rf"Distribución Normal $N(\mu={mu:.1f},\,\sigma^2={sigma**2:.1f})$"
    )
    plt.grid(True)
    plt.show()


widgets.interact(
    dist_normal,

    mu=widgets.FloatSlider(
        value=0,
        min=-5,
        max=5,
        step=0.5,
        description="μ"
    ),

    sigma=widgets.FloatSlider(
        value=1,
        min=0.1,
        max=5,
        step=0.1,
        description="σ"
    ),
)
```

# 5. Distribución Weibull

Esta distribución es muy utilizada para modelar tiempos de vida, fallos y desgaste de componentes, especialmente cuando la tasa de fallos puede variar a medida que transcurre el tiempo.

Esta distribución utiliza como parámetros el coeficiente de forma $k$, que determina el comportamiento de la tasa de fallos, y el parámetro de escala $\lambda$, que determina la escala temporal característica de la distribución.

Ejemplo: Un componente con $k = 2$ y $\lambda = 1000\ [h]$ presenta una tasa de fallos creciente con el tiempo, siendo $\lambda$ una medida de la escala temporal asociada a su vida útil.

```python
def dist_weibull(k: float, lambd: float):
    if k <= 0:
        print("Error: k debe ser mayor que 0.")
        return

    if lambd <= 0:
        print("Error: lambda debe ser mayor que 0.")
        return

    # Rango de valores
    x = np.linspace(0.001, 5*lambd, 1000)

    # Función de densidad Weibull
    y = (
        (k / lambd)
        * (x / lambd)**(k - 1)
        * np.exp(-(x / lambd)**k)
    )

    plt.plot(x, y)



    plt.xlabel("x")
    plt.ylabel("f(x)")
    plt.xlim(0, 15)
    plt.title(
        rf"Distribución Weibull $(k={k:.1f},\,\lambda={lambd:.1f})$"
    )

    plt.grid(True)
    plt.show()


widgets.interact(
    dist_weibull,

    k=widgets.FloatSlider(
        value=1,
        min=0.2,
        max=5,
        step=0.1,
        description="k"
    ),

    lambd=widgets.FloatSlider(
        value=1,
        min=0.5,
        max=5,
        step=0.1,
        description="λ"
    ),
)
```
